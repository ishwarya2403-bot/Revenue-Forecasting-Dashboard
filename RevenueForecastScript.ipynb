```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')

df=pd.read_csv('Revenue Forecast.csv')
print("Dataset\n")
display(df)
```

```python
#checking dataframe shape
print("Number of Rows and Columns in Dataset")
print(df.shape)
```

```python
#checking datatype of dataframe
print("Data types of each columns")
print(df.dtypes)
```

```python
df["Order Date"] = pd.to_datetime(
    df["Order Date"],
    errors="coerce") #Changing and handling Date column
print(df["Order Date"].dtype) 
df["Ship Date"] = pd.to_datetime(
    df["Ship Date"],
    errors="coerce") #Changing and handling Date column
print(df["Ship Date"].dtype)
```

```python
#Rename Sales name
df.rename(columns={
    'Sales':'Revenue With Discount',
    }, inplace=True)
print("Successfully Renamed Columns")
print(df.columns.tolist())
```

```python
df["Unit Price with Discount"] = df["Revenue With Discount"] / df["Quantity"]
df["Unit Price with Discount"] = df["Unit Price with Discount"].round(2)
df["Unit Price without Discount"] = df["Revenue With Discount"] / (df["Quantity"] * (1 - df["Discount"]))
df["Unit Price without Discount"] = df["Unit Price without Discount"].round(2)
df["Revenue Without Discount"] = df["Unit Price without Discount"] * df["Quantity"]
df["Revenue Without Discount"] = df["Revenue Without Discount"].round(2)
print("Column Added Successfully")
df.columns.tolist()
```

```python
#Remove unwanted columns
df.drop(["Order ID"],axis=1,inplace=True,errors="ignore")
print("Sucessfully removed unwanted columns")
```

```python
#exporting the csv file
df.to_csv('RevenueForecastFinalDataset.csv', index=False)
print("File Saved")
```
